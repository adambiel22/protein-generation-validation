````python
import pandas as pd

df = pd.read_csv(
    "diversity_clusters_cluster.tsv",
    sep="\t",
    names=["representative", "member"],
)

n_clusters = df["representative"].nunique()
n_structures = df["member"].nunique()

diversity = n_clusters / n_structures

print("Structures:", n_structures)
print("Clusters:", n_clusters)
print("Diversity:", diversity)
````
