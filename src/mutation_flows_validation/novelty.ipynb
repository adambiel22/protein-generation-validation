````python
import pandas as pd
````

Perform PDB search with the following command:

```bash
foldseek easy-search \
    results/generated_structures/ \
    /raid/adambiel/data_pdb/pdb \
    results/novelty/results.tsv \
    tmp \
    --alignment-type 1 \
    --exhaustive-search \
    --format-output "query,target,qtmscore,ttmscore"
```

````python

df = pd.read_csv(
    "../../results/novelty_ results.tsv",
    sep="\t",
    names=["query", "target", "qtmscore", "ttmscore"],
)

pdb_tm = (
    df.groupby("query")["qtmscore"]
    .max()
    .rename("pdb_tm")
)

result = pdb_tm.to_frame()
result["novel"] = result["pdb_tm"] < 0.5

print(result)
````
