# UniRef50

```python
from datasets import load_dataset, Dataset

ds = load_dataset(
    "ConvergeBio/uniref50",
    split="train",
    streaming=True,
)

proteins = []

for example in ds:
    sequence = example["sequence"]

    if 10 <= len(sequence) <= 1024:
        proteins.append(example)

    if len(proteins) == 500:
        break

dataset = Dataset.from_list(proteins)

dataset.save_to_disk("uniref50_500")
```
